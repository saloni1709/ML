```python
from sklearn.impute import SimpleImputer
# SimpleImputer ko import karte hain (missing values handle karne ke liye)
```

```python
# 👉 Step 1: mean nikalo (average)
# 👉 Step 2: jaha NaN hai → usko mean se fill karo
# [10, 20, NaN, 40]
# Mean = (10+20+40)/3 = 23.3
# After -> [10, 20, 23.3, 40]
si = SimpleImputer(strategy="mean")

# Selected columns me:
# 1. mean (average) calculate karta hai (fit)
# 2. NaN values ko us mean se replace karta hai (transform)
# aur result return karta hai (dataframe ko directly change nahi karta)
ar = si.fit_transform(df[['CWUR_score', 'Minimum_IELTS_score', 'Latitude', 'Longitude']])
```

```python
# ar (NumPy array) ko DataFrame me convert kar rahe hain
# aur df ke float64 columns ke naam assign kar rahe hain
new_df = pd.DataFrame(ar, columns=df.select_dtypes(include="float64").columns)
```

```python
new_df.isnull().sum()    # # har column me kitni missing values (NaN) hain, wo count karta hai
```

# One Hot Encoding

```python
# 👉 Jab data me categorical values (text) hoti hain
# 👉 Unhe numbers me convert karte hain → ML ke liye

# 🔥 Example:

# Before:
# Color
# Red
# Blue
# Green

# After (One Hot Encoding):
# Red   Blue   Green
# 1     0      0
# 0     1      0
# 0     0      1

# 👉 Har category ka alag column ban gaya
```

```python
en_data = df[["University_name", "Region"]]
# 2 columns ka chhota dataset banaya (subset of df)
```

```python
pd.get_dummies(en_data)
# en_data ke categorical columns ko One Hot Encoding me convert karta hai
# har unique value ka alag column banata hai (0/1 ya True/False me)
```

```python
# One Hot Encoding karke naye columns banata hai
# phir .info() un columns ka structure dikhata hai (datatype, count etc.)

pd.get_dummies(en_data).info()
```

```python
sk.preprocessing.OneHotEncoder()
```

```python
from sklearn.preprocessing import OneHotEncoder
```

```python
ohe = OneHotEncoder()
ohe.fit_transform(en_data)

# 👉 fit = "data ko samjho (kaun-kaun si categories hain)"
# 👉 transform = "unhe numbers (0/1) me badlo"

# Sparse matrix output hai (memory efficient format)
# shape (131, 143) → 131 rows aur 143 encoded columns
# 262 stored elements → sirf non-zero values (1s) store ho rahi hain
# baaki sab 0 values store nahi hoti (space bachane ke liye)
```

```python
ohe = OneHotEncoder()
ohe.fit_transform(en_data).toarray()
# .toarray() → sparse output ko normal array me convert karta hai (readable banane ke liye)
```
