# Cost Function

```python
# 👉 Cost Function = Error Measure

# 👉 Ye compare karta hai:
# Actual value (y)
# Predicted value (ŷ)

# ➡️ Ye batata hai:
# Model kitna galat predict kar raha hai

# 🔹 Simple Real-Life Example
# Suppose:

# Actual Salary	   Predicted Salary
# 100	               90
# 200	               180

# Error:
# 100 → 90 → error = 10
# 200 → 180 → error = 20

# 👉 Cost function in errors ko combine karta hai

# Goal of Model :
# 👉 Cost ko minimum karna
# ➡️ Isliye model training hoti hai
# ➡️ Isko hi optimization bolte hain
```

```python
# Types of Cost Function :- 

# 1. Regression Cost Function
# 👉 Used when output is continuous
# Example:
# salary, price, marks

# --------------------------------------------------

# (i) MSE (Mean Squared Error)
# 👉 Most common
# 👉 Mean of squared difference between actual and predicted values

# Formula:
# MSE = (1/n) * Σ (y - ŷ)²

# 👉 Also called L2 Loss
# 👉 Large errors get high punishment (outliers affect more)

# --------------------------------------------------

# (ii) RMSE (Root Mean Squared Error)
# 👉 Square root of MSE

# Formula:
# RMSE = √((1/n) * Σ (y - ŷ)²)

# 👉 Same unit as output (easy to understand)
# 👉 Penalizes large errors (but less than MSE)

# --------------------------------------------------

# (iii) MAE (Mean Absolute Error)
# 👉 Mean of absolute difference

# Formula:
# MAE = (1/n) * Σ |y - ŷ|

# 👉 No square → simple
# 👉 Less sensitive to outliers (robust)

# --------------------------------------------------

# (iv) R² Score (Coefficient of Determination)
# 👉 Measures how well model fits data

# 1 → perfect
# 0 → no learning
# < 0 → very bad model ❌

# --------------------------------------------------

# Summary:
# MSE  → squared error (penalizes big errors)
# RMSE → square root (same unit)
# MAE  → absolute error (robust)
# R²   → model performance

# --------------------------------------------------


# 2. Classification Cost Function
# 👉 Used when output is categorical
# Example:
# 0/1, spam/not spam, cat/dog

# --------------------------------------------------

# (i) Binary Classification
# 👉 Only 2 classes

# Cost Function:
# 👉 Binary Cross Entropy (Log Loss)

# 👉 Wrong prediction → high cost ❌
# 👉 Correct prediction → low cost ✅

# --------------------------------------------------

# (ii) Multi-class Classification
# 👉 More than 2 classes
# Example:
# cat/dog/cow, red/blue/green

# Cost Function:
# 👉 Categorical Cross Entropy

# 👉 Used when multiple classes present
```

# Best Fit Line

```python
# 👉 Best Fit Line ka use hota hai prediction ke liye

# 🔹 Simple Meaning
# Best fit line = wo line jo data ko best represent karti hai
# 👉 Uske baad hum use karte hain:
# ➡️ future values predict karne ke liye


# --------------------------------------------------

# 🔹 Best Fit Line in Linear Regression

# Goal:
# Find a line that best represents the relationship between input (x) and output (y)

# Equation of line:
# y = mx + c
# m → slope
# c → intercept

# --------------------------------------------------

# 🔹 Problem:
# All data points do not lie exactly on the line
# So, there is some error

# Error = (Actual value - Predicted value)
# Error = (y - ŷ)

# --------------------------------------------------

# 🔹 Cost Function (MSE):
# Used to measure total error of model

# Formula:
# Cost = (1/n) * Σ (y - ŷ)^2

# where:
# y  → actual value
# ŷ → predicted value
# n  → number of data points

# --------------------------------------------------

# 🔹 Objective:
# Minimize the cost function

# Best fit line = line with minimum cost

# --------------------------------------------------

# 🔹 How to find best fit line?

# Using Gradient Descent Algorithm

# Steps:
# 1. Initialize m and c with random values
# 2. Calculate predicted values (ŷ)
# 3. Compute cost (MSE)
# 4. Update m and c to reduce cost
# 5. Repeat until cost is minimum

# --------------------------------------------------

# 🔹 Important Points:

# - Lower cost → better model
# - Minimum cost → best fit line
# - Cost function graph is U-shaped (convex)
# - Gradient descent finds lowest point of curve

# --------------------------------------------------

# 🔹 One-line Definition:

# Best fit line is obtained by minimizing the cost function using gradient descent.



# 🔹 Gradient Descent

# Definition:
# Gradient Descent is an optimization algorithm used to minimize the cost function
# by updating model parameters (m and c)

# --------------------------------------------------

# 🔹 Goal:
# Find values of m and c such that cost (error) is minimum

# --------------------------------------------------

# 🔹 Idea:
# Move step-by-step towards minimum point of cost function

# Like going downhill ⛰️
```

# L1(Lasso Regularization), L2(Ridge Regularization)

```python
# 👉 Jab model overfit ho jata hai, tab hum usko control karte hain Is process ko Regularization bolte hain

# # -----------------------------------------------------------------------------------------------------------------

# Overfitting kya hota hai?
# 👉 Over + Fitting = Zyada fit ho jana

# 💡 Matlab:
# Model training data ko itna yaad kar leta hai ki:
# Train data par → 100% sahi
# New data par → galat


# # -------------------------------------------------------------------------------------------------------------

# Formula idea (common)
# Cost function me ek extra term add hoti hai:

# 👉 Cost = Loss + Penalty

# Loss = error (actual vs predicted)
# Penalty = coefficients ko control karne ka term

# # ---------------------------------------------------------------------------------------------------------------------------

# 🔹 L1 (Lasso) – Easy

# 👉 Formula:
# Cost = Loss + λ∑|w|

# 👉 Kya karta hai?
# Kuch weights 0 kar deta hai

# 👉 Result:
# Unnecessary features remove ho jate hain

# 👉 Yaad trick:
# 👉 L1 = Delete features

# # -----------------------------------------------------------------------------------------------------------------------

# L2 (Ridge) – Easy

# 👉 Formula:
# Cost = Loss + λ∑w²

# 👉 Kya karta hai?
# Weights chhote karta hai (0 nahi)

# 👉 Result:
# Sab features rehte hain
# Model stable hota hai

# 👉 Yaad trick:
# 👉 L2 = Reduce impact


# # ------------------------------------------------------------------------------------------------------------

# | L1 (Lasso)                   | L2 (Ridge)          |
# | ---------------------------- | ------------------- |
# | Feature hata deta hai        | Feature nahi hatata |
# | Coefficient = 0 ho sakta hai | 0 nahi hota         |
# | Sparse model                 | Dense model         |

# Sparse = jisme zyada values 0 ho
# Dense = jisme zyada values non-zero ho
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
```

```python
df = pd.read_csv(r"C:\Users\harsh\Downloads\archive\house_price_regression_dataset.csv")
```

```python
df.head(3)
```

```python
print(plt.colormaps())
# matplotlib me available saare colormaps (color styles) ki list print karta hai
```

```python
plt.figure(figsize=(15,10))
sns.heatmap(data=df.corr(), annot=True, cmap="viridis")   # Heatmap bana rahe hain correlation matrix ka
plt.show()

# annot = heatmap ke andar values (numbers) show karta hai
# cmap = heatmap ka color style/palette decide karta hai
```

```python
# X = input features (independent variables)
x = df.iloc[:, :-1]

# iloc = position-based indexing
# :     = sab rows select karo
# :-1   = last column ko chhod kar baaki sab columns lo

# x me saare input features aa gaye (target remove ho gaya)


# y = target / output variable (dependent variable)
y = df["House_Price"]

# House_Price column ko target bana rahe hain
```

```python
sc = StandardScaler()
sc.fit(x)
x = pd.DataFrame(sc.transform(x), columns=x.columns)
```

```python
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)
```

```python
from sklearn.linear_model import LinearRegression, Lasso, Ridge

# 👉 LinearRegression = simple linear model (no regularization)
# 👉 Lasso = L1 regularization (some features ko 0 bana deta hai → feature selection)
# 👉 Ridge = L2 regularization (coefficients ko reduce karta hai, but 0 nahi karta)
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error

# 👉 mean_absolute_error (MAE) = average absolute error (simple difference)
# 👉 mean_squared_error (MSE) = squared error (badi galtiyon ko zyada punish karta hai)

# 👉 Models = prediction ke liye
# 👉 Metrics = performance check karne ke liye
```

# Linear Regression

```python
lr = LinearRegression()
lr.fit(x_train, y_train)
lr.score(x_test, y_test)*100
```

```python
print(mean_squared_error(y_test, lr.predict(x_test)))
# 👉 MSE (Mean Squared Error) print kar rahe hain
# 👉 (actual - predicted)^2 ka average
# 👉 Badi errors ko zyada importance deta hai (penalty high)

print(mean_absolute_error(y_test, lr.predict(x_test)))
# 👉 MAE (Mean Absolute Error)
# 👉 |actual - predicted| ka average
# 👉 Simple aur easy samajhne wala error

print(np.sqrt(mean_squared_error(y_test, lr.predict(x_test))))
# 👉 RMSE (Root Mean Squared Error)
# 👉 MSE ka square root
# 👉 Same unit me error deta hai (target jaisa)
```

```python
lr.coef_
```

```python
plt.figure(figsize=(15,15))
# 👉 Graph ka size set kar rahe hain (bada aur clear view ke liye)

plt.bar(x.columns, lr.coef_)
# 👉 Bar chart bana rahe hain
# 👉 x-axis = feature names (columns)
# 👉 y-axis = unke coefficients (importance)

plt.title("LinearRegression")
# 👉 Graph ka title set kiya

plt.xlabel("columns")
# 👉 x-axis label (features ke naam)

plt.ylabel("coef")
# 👉 y-axis label (coefficient values)

plt.show()
# 👉 Graph display karne ke liye
```

# Lasso

```python
la = Lasso(alpha=0.5)
# 👉 Lasso model banaya
# 👉 alpha = regularization strength (penalty control karta hai)
# 👉 alpha bada → zyada penalty → zyada coefficients 0 ho sakte hain

la.fit(x_train, y_train)
# 👉 Model ko training data par train kiya
# 👉 x_train → input features
# 👉 y_train → target

la.score(x_test, y_test)*100
# 👉 Model ki accuracy check kar rahe hain
# 👉 Ye R² score deta hai (coefficient of determination)
# 👉 *100 → percentage me convert kar diya
```

```python
print(mean_squared_error(y_test, la.predict(x_test)))
print(mean_absolute_error(y_test, la.predict(x_test)))
print(np.sqrt(mean_squared_error(y_test, la.predict(x_test))))
```

```python
plt.figure(figsize=(15,15))
plt.bar(x.columns, la.coef_)
plt.title("Lasso")
plt.xlabel("columns")
plt.ylabel("coef")
plt.show()
```

# Ridge

```python
ri = Ridge(alpha = 10)
ri.fit(x_train, y_train)
ri.score(x_test, y_test)*100
```

```python
print(mean_squared_error(y_test, ri.predict(x_test)))
print(mean_absolute_error(y_test, ri.predict(x_test)))
print(np.sqrt(mean_squared_error(y_test, ri.predict(x_test))))
```

```python
plt.figure(figsize=(15,15))
plt.bar(x.columns, ri.coef_)
plt.title("Ridge")
plt.xlabel("columns")
plt.ylabel("coef")
plt.show()
```

```python
df = pd.DataFrame({"col_name":x.columns, "LinearRegression":lr.coef_, "Lasso":la.coef_, "Ridge":ri.coef_})
```
