# CLASSIFICATION ALGORITHM

```python
# Classification algorithm ka use new data ko categories me divide karne ke liye hota hai

# Model training data se seekhta hai
# Phir new data ko kisi class/group me assign karta hai

# Examples:
# Yes/No  
# 0/1
# Spam / Not Spam
# Cat / Dog

# Classes ko target / label bhi bolte hain
```

# TYPES OF CLASSIFICATION

```python
# 1. Binary Classification
# Sirf 2 output hote hain

# Examples:
# Spam / Not Spam
# Pass / Fail
# 0 / 1


# 2. Multi-class Classification
# 2 se zyada output hote hain

# Examples:
# Crop types
# Music categories
# Animal types
```

# TYPES OF ML (CLASSIFICATION ALGORITHMS)

```python
# Non-linear Models
# Data straight line se separate nahi hota

# Examples:
# KNN (K-Nearest Neighbors)
# SVM
# Naive Bayes
# Decision Tree
# Random Forest


# Linear Models
# Data ko straight line se separate karte hain

# Examples:
# Logistic Regression
# SVM (linear case)
```

# EVALUATING CLASSIFICATION MODEL

```python
# We check how well the model is performing using different metrics.

# 1. Log Loss (Cross-Entropy Loss)
# Measures how wrong the predictions are
# Lower value = better model
# It considers probabilities, not just correct/incorrect

# 👉 Example:
# If model says 0.9 (very confident) but wrong → high loss

# 2. Confusion Matrix
# A table that shows correct and incorrect predictions

# It has 4 parts:

# True Positive (TP) → Correct Yes
# True Negative (TN) → Correct No
# False Positive (FP) → Wrong Yes
# False Negative (FN) → Wrong No

# 👉 Helps to understand where model is making mistakes

# 3. AUC-ROC Curve
# Measures model performance on a scale of 0 to 1
# 1 → Perfect model
# 0.5 → Random guess
# Closer to 1 → Better

# 👉 Shows how well model separates classes

# Log Loss → Error measure (lower is better)
# Confusion Matrix → Detailed prediction results
# AUC-ROC → Overall performance (higher is better)
```

```python
# Logistic Regression ek supervised ML algorithm hai

# Use:
# Categorical output predict karta hai

# Output:
# Direct 0/1 nahi deta
# Probability deta hai (0 se 1 ke beech)

# Example:
# 0.8 -> Spam
# 0.2 -> Not Spam
```

# OVR = One vs Rest (One vs All)

```python
# It is used in multi-class classification
# One class is treated as positive (1) and all other classes as negative (0)

# 🔹 How it works?
# If there are 3 classes:
# 👉 Cat, Dog, Sheep

# Then 3 models are created:
# Cat vs (Dog + Sheep)
# Dog vs (Cat + Sheep)
# Sheep vs (Cat + Dog)


# 🔹 How prediction works?
# Each model gives a probability score
# The class with the highest probability is selected


# 🔹 Where is it used?
# Logistic Regression
# SVM
# Other classification algorithms
```

```python
from sklearn.linear_model import LogisticRegression
```

```python
lr = LogisticRegression(multi_class="ovr")
lr.fit(x_train, y_train)
```

```python
lr.score(x_test, y_test)
```
