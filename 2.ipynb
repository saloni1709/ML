```python
df = pd.DataFrame({"name" : ["wscube", "cow", "cat", "dog", "black"]})
df
```

```python
# LabelEncoder import kiya (categorical text data ko numbers me convert karne ke liye)
from sklearn.preprocessing import LabelEncoder
```

```python
# LabelEncoder object banaya (text ko numbers me convert karne ke liye)
le = LabelEncoder()

# 'name' column ki categories seekhta hai (fit)
# aur unhe numbers (0,1,2,...) me convert karta hai (transform)
le.fit_transform(df["name"])
```

```python
# LabelEncoder object banaya (text ko numbers me convert karne ke liye)
le = LabelEncoder()

# 'name' column ko encode karke naya column 'en_name' me store kiya
# fit() → unique names seekhta hai
# transform() → unhe numbers (0,1,2,...) me convert karta hai
df["en_name"] = le.fit_transform(df["name"])
```

# Ordinal Encoding

```python
# 👉 Jab categories ka order (ranking) hota hai
# 👉 tab unhe numbers me convert karte hain order ke hisaab se

# 🔍 Example:

# Data:
# Size → Small, Medium, Large

# Small < Medium < Large

# Encoding:
# Small  → 0  
# Medium → 1  
# Large  → 2
```

```python
# ❗ Difference (important)

#   Encoding	             Kab use kare
# LabelEncoder	      order matter nahi karta
# OneHotEncoder	      categories independent ho
# OrdinalEncoder	      order important ho
```

```python
df = pd.DataFrame({"Size": ["s", "m", "l", "xl", "s", "m", "l", "s", "s", "l", "xl", "m"]})
df.head()
```

```python
ord_data = [["s", "m", "l", "xl"]]
```

```python
from sklearn.preprocessing import OrdinalEncoder
```

```python
oe = OrdinalEncoder(categories=ord_data)  
# ordered categories define karke encoder object banaya

oe.fit(df[["Size"]])  
# Size column ki categories ko learn (mapping banata hai) kar raha hai
```

```python
oe.transform(df[["Size"]])
```

```python
df["Size_en"] = oe.transform(df[["Size"]])
df
```

```python
ord_data1 = {"s":5, "m":6, "l":7, "xl":8}
```

```python
df["size_en_map"] = df["Size"].map(ord_data1)
```
