# Handling Duplicate Data

```python
import pandas as pd
```

```python
data = {"name":["a","b","c","d","a","c"], "eng":[8,7,5,8,8,5], "hindi":[2,3,4,5,2,6]}
```

```python
df = pd.DataFrame(data)
df
```

```python
# df["duplicated"] = df.duplicated()  

# # jo col. duplicate hoga -> True ,  else -> False
```

```python
df
```

```python
df.drop_duplicates()   
# duplicate rows remove karta hai
# default: first occurrence ko rakhta hai
# original df change nahi hota (temporary output)
```

```python
df.drop_duplicates(inplace=True)   # duplicate value (col. -> 4) ko dlt  [Original dataset ko change]
```
