```python
from sklearn.preprocessing import OneHotEncoder
```

```python
# Create One Hot Encoder
# drop='first' removes the first category to avoid dummy variable trap
# sparse_output=False returns a normal NumPy array instead of a sparse matrix
# dtype=np.int32 stores encoded values as integers
ohe = OneHotEncoder(
    drop='first',
    sparse_output=False,
    dtype=np.int32
)
```

```python
X_train_new = ohe.fit_transform(X_train[['gender']])
```

# 4. OneHotEncoding with Top Categories

```python
counts = df['gender'].value_counts()
```

```python
df['gender'].nunique()
```

```python
# Threshold = Limit / Cut-off Value
# Aap decide karte ho ki kis frequency ke niche wali categories ko rare category maana jayega.

threshold = 5       # Jinki frequency 5 ya usse kam hai, unhe rare category maan lo.

repl = counts[counts <= threshold].index    # # Get categories whose count is 5 or less
```

```python
# Replace rare categories with 'uncommon' (a label used for all low-frequency values)
# Example: if 'Other', 'Unknown', 'NA' occur very few times, they are replaced with 'uncommon'
# Then convert categories into 0/1 columns using one-hot encoding
# Finally, display 5 random rows

pd.get_dummies(df['gender'].replace(repl, 'uncommon')).sample(5)
```
