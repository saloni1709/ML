# Feature Selection Techniques :-
### 1. Backward Elimination (Using MLXTEND) 
### 2. Forward Elimination (Using MLXTEND)

```python
# Feature Selection Techniques
# Model ke liye important features select karna aur useless hata dena

# 1. Forward Elimination :- 
# 👉 Start with zero features

# 👉 Phir:
# ek-ek karke features add karte hain
# jo model performance improve kare

# 🔁 Steps
# Start: ❌ no features
# Har feature ko try karo
# Jo best performance de → add karo
# Repeat until best set mil jaye

# 👉 Empty se start → useful features add karo


# 2. Backward Elimination
# 👉 Start with all features

# 👉 Phir:
# ek-ek karke least useful remove

# 🔁 Steps
# Start: ✅ all features
# Har feature ka impact check karo
# Jo useless hai → remove
# Repeat

# 👉 Sabse start → useless hatao


# | Feature  | Forward                      | Backward                   |
# | -------- | ---------------------------- | -------------------------- |
# | Start    | No features                  | All features               |
# | Process  | Add                          | Remove                     |
# | Speed    | Slow (jab features zyada ho) | Fast (small dataset me)    |
# | Use case | Jab features bahut zyada ho  | Jab features manageable ho |
```

```python
# 👉 MLXTEND = Machine Learning Extensions

# 👉 Ye ek Python library hai jo:
# scikit-learn ko extend karti hai (extra tools provide karti hai)

# 🔑 Main Tool
# 🔹 Sequential Feature Selector (SFS)
# 👉 It performs:
# Forward Selection (adds features step by step)
# Backward Elimination (removes features step by step)

# 👉 sklearn basic ML tools deta hai
# 👉 mlxtend → extra advanced utilities deta hai

# 🔹 MLXTEND me kya milta hai?
# Feature Selection (Forward / Backward) ✅
# Frequent Pattern Mining (Apriori)
# Ensemble methods
# Evaluation tools

# 👉 MLXTEND = extra ML tools for feature selection & more
```

```python
from sklearn.linear_model import LinearRegression
from mlxtend.feature_selection import SequentialFeatureSelector
```

```python
lr = LinearRegression()
```

```python
fs = SequentialFeatureSelector(lr, k_features=5, forward=True)   # Forward ke liye forward=True,  Forward ke liye backward=False
# k_features=5 → final me 5 best features select karega
```

```python
fs.fit(x, y)
```

```python
fs.feature_names
```

```python
fs.k_feature_names_
```
