# Working with Dates

```python
# Converting to datetime datatype
date['date'] = pd.to_datetime(date['date'])
```

```python
date.info()
```

### 1. Extract year

```python
date['date_year'] = date['date'].dt.year

date.sample(5)
```

### 2. Extract Month

```python
date['date_month_no'] = date['date'].dt.month

date.sample(5)
```

```python
date['date_month_name'] = date['date'].dt.month_name()

date.sample(5)
```

### 3. Extract Days

```python
date['date_day'] = date['date'].dt.day

date.sample(5)
```

#### day of week

```python
date['date_dow'] = date['date'].dt.dayofweek
date.sample(5)
```

####  day of week - name

```python
date['date_dow_name'] = date['date'].dt.day_name()

date.sample(5)
```

#### is weekend?

```python
date['date_is_weekend'] = np.where(date['date_dow_name'].isin(['Sunday', 'Saturday']), 1,0)

date.drop(columns=['product_id','city_id','orders']).head()
```

#### Extract Quarter

```python
date['quarter'] = date['date'].dt.quarter

date.drop(columns=['product_id','city_id','orders']).head()
```

#### Extract Semester

```python
date['semester'] = np.where(date['quarter'].isin([1,2]), 1, 2)

date.drop(columns=['product_id','city_id','orders']).head()
```

```python
# Convert the 'date' column from string/object format to datetime format
# This allows us to perform time-based operations like extracting year, month, day,
# and calculating time differences between dates
time['date'] = pd.to_datetime(time['date'])
```

```python
time.info()
```

```python
time.head()
```

# Extract Time part

```python
time['time'] = time['date'].dt.time
```

```python
time.head()
```

### Time difference

```python
today - date['date']
```

#### in Seconds

```python
# Calculate time difference between 'today' and each date in the column
# (today - time['date']) gives a timedelta (difference in days/hours/etc.)

# Convert that timedelta into seconds by dividing with 1 second unit
# np.timedelta64(1, 's') means 1 second, so result becomes total seconds difference

(today - time['date']) / np.timedelta64(1, 's')
```

#### in minutes

```python
(today - time['date'])/np.timedelta64(1,'m')
```

#### in hours

```python
(today - time['date'])/np.timedelta64(1,'h')
```
