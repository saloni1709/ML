```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('titanic.csv')
```

```python
df.head()
```

```python
df['number'].unique()
```

```python
fig = df['number'].value_counts().plot.bar()
fig.set_title('Distribution of Numbers')
```

```python
# Convert the 'number' column into numeric format and store it as a new column
# errors='coerce' will convert invalid values to NaN
# downcast='integer' reduces memory usage by converting to a smaller integer type if possible

df['number_numerical'] = pd.to_numeric(df['number'], errors='coerce', downcast='integer')
```

```python
# extract categorical part

df['number_categorical'] = np.where(df['number_numerical'].isnull(), df['number'], np.nan)
```

```python
df.head()
```

```python
df['Cabin'].unique()
```

```python
df['Ticket'].unique()
```

```python
# Extract numeric part from 'Cabin' column using regex
# \d+ means: one or more digits (0-9)
# Example: 'C123' -> 123
df['cabin_num'] = df['Cabin'].str.extract('(\d+)')

# Extract categorical part from 'Cabin' column
# str[0] means first character of the string (index 0)
# Example: 'C123' -> 'C'
df['cabin_cat'] = df['Cabin'].str[0]
```

```python
df.head()
```

```python
df['cabin_cat'].value_counts().plot(kind='bar')
```

```python
# Extract the last part of 'Ticket' as numeric value
# Step 1: split ticket string by spaces and take last part
# Step 2: convert it into numeric (invalid values become NaN)
df['ticket_num'] = df['Ticket'].apply(lambda s: s.split()[-1])
df['ticket_num'] = pd.to_numeric(
    df['ticket_num'],
    errors='coerce',
    downcast='integer'
)

# Extract the first part of 'Ticket' as category
# Step 1: split ticket string and take first part
df['ticket_cat'] = df['Ticket'].apply(lambda s: s.split()[0])

# If first part is purely numeric, convert it to NaN (not a valid category)
df['ticket_cat'] = np.where(
    df['ticket_cat'].str.isdigit(),
    np.nan,
    df['ticket_cat']
)

# Display first 5 rows of dataframe
df.head()
```
