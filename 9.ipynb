# Polynomial Regression

```python
# 𝑦 = 𝑏0 + 𝑏1𝑥 + 𝑏2𝑥^2 + 𝑏3𝑥^3 + ⋯ + 𝑏𝑛𝑥^𝑛

# 👉 It is used when data is not linear (not straight line)
# 👉 Instead of line → it fits a curve

# Difference
# Linear Regression → straight line
# Polynomial Regression → curved line 📈

# 👉 Salary vs Experience
# If straight line fit nahi ho rahi
# Then use polynomial → better fit
```

```python
# | Feature           | Linear Regression | Multiple Linear Regression     | Polynomial Regression       |
# | ----------------- | ----------------- | ------------------------------ | --------------------------- |
# | **No. of Inputs** | 1 feature         | 2 or more features             | 1 (or more) but with powers |
# | **Relationship**  | Straight line     | Plane / multi-dimensional line | Curve (non-linear)          |
# | **Equation**      | y = mx + c        | y = b₀ + b₁x₁ + b₂x₂ + ...     | y = b₀ + b₁x + b₂x² + ...   |
# | **Graph Shape**   | Straight line 📈  | Multi-dimensional              | Curved line 📊              |
# | **Use Case**      | Simple prediction | Real-world (many factors)      | Non-linear data             |
# | **Complexity**    | Low               | Medium                         | High                        |
```
