```python
import numpy as np
```

```python
speed = [10, 20, 30, 40, 50]
x = np.mean(speed)
x
```

```python
print(x)
```

```python
x1 = np.median(speed)
print(x1)
```

```python
distance = [10, 20, 40, 50]
x2 = np.median(distance)
print(x2)
```

```python
from scipy import stats
```

```python
x3 = stats.mode(speed)   # NumPy does not have a mode() function.
print(x3)
```

```python
x4 = np.std(distance)
print(x4)
```

```python
x5 = np.var(distance)
print(x5)
```
