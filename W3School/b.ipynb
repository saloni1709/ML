# Data Distribution (Probability Distribution)

### Uniform Distribution

```python
# np.random.uniform(low, high, size)
```

```python
# Generate 10 random decimal (float) numbers
# between 0 and 1 from a Uniform Distribution.
# 0 is inclusive(included), 1 is exclusive(not included).
# Every number in this range has an equal chance of being selected.

x = np.random.uniform(0, 1, 10)

# Display the generated random numbers
x
```

### Histogram

```python
import matplotlib.pyplot as plt
```

```python
# Create a histogram of the data stored in x
# '5' means divide the data into 5 bins (intervals)

plt.hist(x, 5)

# Display the histogram
plt.show()
```

### Big Data Distributions

```python
# Generate 100000 random decimal numbers
# from a Uniform Distribution.
# Range: 0.0 (inclusive) to 5.0 (exclusive)

x1 = np.random.uniform(0.0, 5.0, 100000)

# Plot a histogram of the generated data.
# Divide the range into 100 bins (intervals).

plt.hist(x1, bins=100)

# Display the histogram.

plt.show()
```

## Normal Data Distribution

```python
# np.random.normal(mean, std, size)
```

```python
# Generate 100000 random numbers
# from a Normal (Gaussian) Distribution.

# Mean (μ) = 0
# Standard Deviation (σ) = 1
# Size = 100000 random values

y = np.random.normal(0, 1, 100000)

# Display the generated random numbers
y
```

```python
plt.hist(y, 100)
plt.show()
```
