# Scale

```python
import numpy as np
import pandas as pd
```

#### The standardization(Z-score Normalization) method uses this formula:

## z = (x - u) / s

#### Where z is the new value, x is the original value, u is the mean and s is the standard deviation.

```python
df = pd.DataFrame({
    "Weight": [790,1160,929,865,1140],
    "Volume": [1000,1200,1000,900,1500],
    "CO2": [99,95,95,90,105]
})
```

```python
from sklearn.preprocessing import StandardScaler

sc = StandardScaler()
```

```python
X = df[['Weight', 'Volume']]
scaledX = sc.fit_transform(X)
print(scaledX)
```

```python
# Predict CO2 Values using Linear Regression with feature scaling

import pandas as pd

# Input features (independent variables)
# Weight and Volume are used to predict CO2 emission
X = df[['Weight', 'Volume']]

# Target variable (dependent variable)
# CO2 emission is what we want to predict
y = df['CO2']

# Apply Standard Scaling (Z-score normalization)
# Converts features to same scale (mean=0, std=1)
scaledX = sc.fit_transform(X)

from sklearn.linear_model import LinearRegression

# Create Linear Regression model
lr = LinearRegression()

# Train model on scaled features and target
lr.fit(scaledX, y)
    
# New input data (Weight=1000, Volume=1200)
# IMPORTANT: must use SAME scaler used on training data
scaled = sc.transform([[1000, 1200]])

# Predict CO2 using trained model
# scaled[0] converts 2D array to 1D format
y_pred = lr.predict([scaled[0]])

# Print predicted CO2 emission
print(y_pred)
```
