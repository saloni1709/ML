# Decision Tree

```python
import pandas as pd
```

```python
df = pd.DataFrame({
    "Age": [36, 42, 23, 52, 43, 44, 66, 35, 52, 35, 24, 18, 45],
    "Experience": [10, 12, 4, 4, 21, 14, 3, 14, 13, 5, 3, 3, 9],
    "Rank": [9, 4, 6, 4, 8, 5, 7, 9, 7, 9, 5, 7, 9],
    "Nationality": ["UK", "USA", "N", "USA", "USA", "UK", "N", "UK", "N", "N", "USA", "UK", "UK"],
    "Go": ["NO", "NO", "NO", "NO", "YES", "NO", "YES", "YES", "YES", "YES", "NO", "YES", "YES"]
})
```

```python
print(df)
```

```python
df['Nationality'] = df['Nationality'].map({
    "UK": 0,
    "USA": 1,
    "N" : 2
})

df['Go'] = df["Go"].map({
    "NO": 0,
    "YES": 1
})
```

```python
df.head()
```

```python
X = df.iloc[:, :-1]
y = df.iloc[:, -1]
```

```python
X
```

```python
y
```

```python
from sklearn.tree import DecisionTreeClassifier
```

```python
import matplotlib.pyplot as plt
```

```python
dt = DecisionTreeClassifier()
dt.fit(X, y)
```

```python
from sklearn import tree      # Decision Tree ko visualize karne ke liye tree module import

# plot_tree() function trained Decision Tree model ko graphical form me display karta hai.
# Isse hum dekh sakte hain:
# - Root node
# - Decision conditions (splits)
# - Leaf nodes
# - Predicted class/output

tree.plot_tree(dt)
```

```python
print(dt.predict([[40, 10, 7, 1]]))
```

```python
print(dt.predict([[40, 10, 6, 1]]))
```

```python
# Create a Decision Tree Classifier model
# criterion='entropy' uses Information Gain to find the best split
# max_depth=3 limits the tree depth to prevent overfitting
# random_state=42 ensures reproducible and consistent results

dtree = DecisionTreeClassifier(
    criterion='entropy',
    max_depth=3,
    random_state=42
)

# Display the model configuration
dtree
```

```python
# Create a Decision Tree Classifier model using Gini Impurity
# Gini measures the impurity of a split and selects the best split
# It is the default criterion used in Decision Trees

DecisionTreeClassifier(criterion='gini')
```
