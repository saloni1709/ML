# Hierarchical Clustering
#### Hierarchical Clustering is an unsupervised learning algorithm
#### used to group similar data points into clusters.

### 1. Agglomerative Clustering
### 2. Dendrogram

```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
x = [4, 5, 10, 4, 3, 11, 14 , 6, 10, 12]
y = [21, 19, 24, 17, 16, 25, 24, 22, 21, 21]

plt.scatter(x, y)
plt.show()
```

```python
# Import dendrogram for visualizing hierarchical cluster relationships
# Import linkage for computing the hierarchical clustering structure
from scipy.cluster.hierarchy import dendrogram, linkage
```

```python
# Dataset

x = [4,5,10,4,3,11,14,6,10,12]
y = [21,19,24,17,16,25,24,22,21,21]
```

```python
# Hierarchical Clustering algorithms expect the input data
# in the form of data points (rows and columns/features).
#
# Currently, we have two separate lists:
# x = [4, 5, 10, ...]
# y = [21, 19, 24, ...]
#
# We use zip() to combine the corresponding values of x and y
# into coordinate pairs, where each pair represents one data point.
#
# Before zip():
# x = [4, 5, 10]
# y = [21, 19, 24]
#
# After zip():
# [(4, 21), (5, 19), (10, 24)]
#
# This converts the data into the format required by
# Hierarchical Clustering algorithms and dendrogram visualization.

data = list(zip(x, y))
```

```python
# Calculate the distances between data points and
# determine how clusters should be merged.
# 'ward' minimizes the variance within clusters.
# 'euclidean' is used to measure the distance.
# The resulting linkage data is used to create a dendrogram.

linkage_data = linkage(
    data,
    method='ward',
    metric='euclidean'
)
```

```python
# Plot dendrogram

dendrogram(linkage_data)
plt.title('Dendrogram')
plt.show()
```

```python
# Import the Agglomerative Clustering algorithm
from sklearn.cluster import AgglomerativeClustering

# Create a Hierarchical Clustering model
# n_clusters=2 means create 2 clusters
# linkage='ward' merges clusters by minimizing variance

hc = AgglomerativeClustering(
    n_clusters=2,
    linkage='ward'
)
```

```python
# Fit and predict clusters

hc.fit_predict(data)
```

```python
# Plot the clustered data points

# x and y are coordinates of data points

# c=hc.labels_ means:
# - hc.labels_ contains cluster numbers (0, 1, 2, ...)
# - each number represents a different cluster
# - matplotlib automatically assigns different colors
#   to different cluster numbers using a color map

# We are NOT giving manual colors like 'red' or 'blue'
# because we want to visualize clusters automatically

plt.scatter(x, y, c=hc.labels_)
plt.show()
```
