# Preprocessing

### Categorical Data
##### Categorical data preprocessing is the process of converting non-numeric categories into numerical values so that machine learning algorithms can process them. Common techniques include Label Encoding and One-Hot Encoding.

```python
import pandas as pd
import numpy as np
```

```python
# Create dataset
df = pd.DataFrame({
    "Car": [
        "Toyota", "Mitsubishi", "Skoda", "Fiat",
        "Mini", "VW", "Skoda", "Mercedes",
        "Ford", "Audi", "Hyundai", "Suzuki"
    ],

    "Model": [
        "Aygo", "Space Star", "Citigo", "500",
        "Cooper", "Up!", "Fabia", "A-Class",
        "Fiesta", "A1", "I20", "Swift"
    ],

    "Volume": [
        1000, 1200, 1000, 900,
        1500, 1000, 1400, 1500,
        1500, 1600, 1100, 1300
    ],

    "Weight": [
        790, 1160, 929, 865,
        1140, 929, 1109, 1365,
        1112, 1150, 980, 990
    ],

    "CO2": [
        99, 95, 95, 90,
        105, 105, 90, 92,
        98, 99, 99, 101
    ]
})

print(df)
```

##### Methods to Convert Categorical Data

# One Hot Encoding

```python
df_encoded = pd.get_dummies(
    df,
    columns = ['Car', 'Model'],
    dtype = int
)

df_encoded
```

# LabelEncoder

```python
from sklearn.preprocessing import LabelEncoder
```

```python
la = LabelEncoder()
```

```python
# Convert the categorical values in the
# 'Car' column into numerical labels

df['Car'] = la.fit_transform(df['Car'])

df['Car']
```

```python
# Convert the categorical values in the
# 'Model' column into numerical labels

df['Model'] = la.fit_transform(df['Model'])

df['Model']
```
