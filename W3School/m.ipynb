# K-means

##### K-means is an unsupervised learning method for clustering data points. The algorithm iteratively divides data points into K clusters by minimizing the variance in each cluster.

```python
import matplotlib.pyplot as plt
```

```python
x = [4, 5, 10, 4, 3, 11, 14 , 6, 10, 12]
y = [21, 19, 24, 17, 16, 25, 24, 22, 21, 21]

plt.scatter(x, y)
plt.show()
```

```python
from sklearn.cluster import KMeans
```

```python
# Convert x and y into coordinate points

data = list(zip(x, y))
```

```python
kmeans = KMeans(
    n_clusters = 2,
    random_state = 42
)
```

```python
kmeans.fit(data)
```

```python
kmeans.labels_
```

### Plot clusters

```python
plt.scatter(
    x,
    y,
    c = kmeans.labels_
)
```

### Centroids

```python
# Display cluster centers
kmeans.cluster_centers_
```

### Predict new point

```python
# Predict which cluster
# the new point belongs to

print(
    kmeans.predict([[8,20]])
)
```

# Elbow Method
### Used to find the best value of K.

```python
# Create an empty list to store WCSS values
# (WCSS = Within-Cluster Sum of Squares)

wcss = []

# Try different numbers of clusters (K)
# from 1 to 10

for i in range(1, 11):

    # Create a K-Means model with i clusters
    # random_state=42 ensures the same result
    # every time the code is executed

    km = KMeans(
        n_clusters=i,
        random_state=42
    )

    # Train the K-Means model on the dataset
    km.fit(data)

    # Store the WCSS (inertia) value
    # for the current number of clusters

    wcss.append(km.inertia_)
```

```python
plt.plot(
    range(1, 11),
    wcss
)
plt.show()
```
