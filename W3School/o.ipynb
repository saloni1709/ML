# Cross Validation

#### ross Validation is a technique used to check how well a machine learning model performs on unseen data.

#### Instead of using only one train-test split:

##### Train = 80%
##### Test  = 20%

#### we split the dataset multiple times and evaluate the model multiple times.

```python
from sklearn import datasets

from sklearn.datasets import load_iris
```

```python
df = load_iris()
```

```python
X = df.data
y = df.target
```

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import KFold, cross_val_score
```

```python
clf = DecisionTreeClassifier(random_state = 42)
```

```python
k_folds = KFold(n_splits = 5)
scores = cross_val_score(clf, X, y, cv = k_folds)
```

```python
print(scores)
print("Mean accuracy:", scores.mean())
print("Standard deviation:", scores.std())
print(len(scores))
```

# Stratified K-Fold

```python
from sklearn.model_selection import StratifiedKFold
```

```python
clf = DecisionTreeClassifier(random_state = 42)

sk_folds = StratifiedKFold(n_splits = 5)

scores = cross_val_score(clf, X, y, cv = sk_folds)
```

```python
print("Cross Validation Scores: ", scores)
print("Average CV Score: ", scores.mean())
print("Number of CV Scores used in Average: ", len(scores))
```

# Leave-One-Out (LOO)

### Har time 1 sample test hota hai, baaki sab training.

#### Total data = 100 samples

##### Run 1 → 1 sample test, 99 train
##### Run 2 → 1 sample test, 99 train
#### Run 100 → 1 sample test, 99 train

```python
from sklearn.model_selection import LeaveOneOut
```

```python
loo = LeaveOneOut()
```

```python
scores = cross_val_score(clf, X, y, cv = loo)
```

```python
print("Cross Validation Scores: ", scores)
print("Average CV Score: ", scores.mean())
print("Number of CV Scores used in Average: ", len(scores))
```

# Leave-P-Out (LPO)

### Har time P samples test hoti hain

#### Example:    P = 2

##### Run 1 → 2 test, rest train
##### Run 2 → 2 test, rest train

```python
from sklearn.model_selection import LeavePOut
```

```python
lpo = LeavePOut(p = 2)
```

```python
scores = cross_val_score(clf, X, y, cv = lpo)
```

```python
print("Cross Validation Scores: ", scores)
print("Average CV Score: ", scores.mean())
print("Number of CV Scores used in Average: ", len(scores))
```

# Shuffle Split

### Randomly data ko split karta hai multiple times.

##### Run 1 → random split
##### Run 2 → random split
##### Run 3 → random split

```python
from sklearn.model_selection import ShuffleSplit

ss = ShuffleSplit(train_size = 0.6, test_size = 0.3, n_splits = 5)
```

```python
scores = cross_val_score(clf, X, y, cv = ss)

print("Cross Validation Scores: ", scores)
print("Average CV Score: ", scores.mean())
print("Number of CV Scores used in Average: ", len(scores))
```
